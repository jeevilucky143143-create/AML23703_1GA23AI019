```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

def full_adder(A, B, Cin):

    qc = QuantumCircuit(5, 5)

    if A == 1:
        qc.x(0)

    if B == 1:
        qc.x(1)

    if Cin == 1:
        qc.x(2)

    qc.cx(0, 3)
    qc.cx(1, 3)
    qc.cx(2, 3)

    qc.ccx(0, 1, 4)
    qc.ccx(0, 2, 4)
    qc.ccx(1, 2, 4)

    qc.measure(0, 0)
    qc.measure(1, 1)
    qc.measure(2, 2)
    qc.measure(3, 3)
    qc.measure(4, 4)

    return qc


simulator = AerSimulator()

print("FULL ADDER RESULTS")
print("=" * 50)

for A in [0, 1]:
    for B in [0, 1]:
        for Cin in [0, 1]:

            qc = full_adder(A, B, Cin)

            job = simulator.run(qc, shots=1)
            result = job.result()
            counts = result.get_counts()

            output = list(counts.keys())[0]

            cout = output[0]
            sum_bit = output[1]

            print(
                f"A={A}, B={B}, Cin={Cin} "
                f"--> Sum={sum_bit}, Cout={cout}"
            )
```
