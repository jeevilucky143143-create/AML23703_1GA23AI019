```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
import numpy as np

theta = np.pi / 3

qc = QuantumCircuit(2)

qc.ry(theta, 0)
qc.cx(0, 1)

print(qc.draw())

state = Statevector.from_instruction(qc)

print("\nStatevector:")
print(state)

print("\nAmplitudes:")

for i, amplitude in enumerate(state.data):

    binary_state = format(i, '02b')

    if abs(amplitude) > 1e-10:
        print(
            f"|{binary_state}> : "
            f"{amplitude.real:.6f} + "
            f"{amplitude.imag:.6f}i"
        )

probabilities = state.probabilities_dict()

print("\nProbabilities:")

for basis_state, probability in probabilities.items():
    print(f"|{basis_state}> : {probability:.6f}")

expected_00 = np.cos(theta / 2) ** 2
expected_11 = np.sin(theta / 2) ** 2

print("\nExpected theoretical probabilities:")
print(f"P(00) = {expected_00:.6f}")
print(f"P(11) = {expected_11:.6f}")

print("\nVerification:")

if (
    np.isclose(probabilities.get('00', 0), expected_00)
    and
    np.isclose(probabilities.get('11', 0), expected_11)
):
    print("SUCCESS!")
    print("The desired arbitrary entangled state was created.")
else:
    print("State verification failed.")
```
