```python
import random
import numpy as np

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit_aer import AerSimulator
```

```python
def show_state(circuit, title):
    state = Statevector.from_instruction(circuit)
    print("="*60)
    print(title)
    print(circuit.draw())
    print("\nStatevector:")
    print(state)
    print("="*60)
    return state
```

```python
qc_x = QuantumCircuit(1)
qc_x.x(0)

state_x = show_state(qc_x, "X Gate")
```

```python
qc_y = QuantumCircuit(1)
qc_y.y(0)

state_y = show_state(qc_y, "Y Gate")
```

```python
qc_z = QuantumCircuit(1)
qc_z.z(0)

state_z = show_state(qc_z, "Z Gate")
```
