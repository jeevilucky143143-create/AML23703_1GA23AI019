```python
import random
import numpy as np

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit_aer import AerSimulator
```

```python
def show_state(circuit, title):
    state = Statevector.from_instruction(circuit)
    print("="*60)
    print(title)
    print(circuit.draw())
    print("\nStatevector:")
    print(state)
    print("="*60)
    return state
```

```python
coin = QuantumCircuit(1)

coin.h(0)

coin.measure_all()

print(coin.draw())
```

```python
sim = AerSimulator()

job = sim.run(coin, shots=1000)

result = job.result()

counts = result.get_counts()

print("Measurement Counts:\n")
print(counts)
```

```python
shots = sum(counts.values())

prob_0 = counts.get('0',0)/shots

prob_1 = counts.get('1',0)/shots

print("Probability of Heads (0):",prob_0)

print("Probability of Tails (1):",prob_1)
```
