```python
import random
import numpy as np

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit_aer import AerSimulator
```

```python
def show_state(circuit, title):
    state = Statevector.from_instruction(circuit)
    print("="*60)
    print(title)
    print(circuit.draw())
    print("\nStatevector:")
    print(state)
    print("="*60)
    return state
```

```python
qc_minus = QuantumCircuit(1)

qc_minus.h(0)
qc_minus.z(0)

state_minus = show_state(qc_minus, "|-> State using H and Z")
```

```python
expected = np.array([1/np.sqrt(2), -1/np.sqrt(2)])

print("Expected State:")
print(expected)

print("\nObtained State:")
print(state_minus.data)

print("\nVerification:")
print(np.allclose(expected, state_minus.data))
```
