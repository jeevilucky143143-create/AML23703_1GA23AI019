```python
import random
import numpy as np

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit_aer import AerSimulator
```

```python
def show_state(circuit, title):
    state = Statevector.from_instruction(circuit)
    print("="*60)
    print(title)
    print(circuit.draw())
    print("\nStatevector:")
    print(state)
    print("="*60)
    return state
```

```python
qc_hzh = QuantumCircuit(1)

qc_hzh.h(0)

qc_hzh.z(0)

qc_hzh.h(0)

state_hzh = show_state(qc_hzh,"HZH Circuit")
```

```python
qc_x2 = QuantumCircuit(1)

qc_x2.x(0)

state_x2 = show_state(qc_x2,"X Gate Circuit")
```

```python
print("HZH State:")
print(state_hzh.data)

print("\nX Gate State:")
print(state_x2.data)

print("\nEquivalent?")

print(np.allclose(state_hzh.data,state_x2.data))
```

```python
print("Analytical Proof")

print()

print("Hadamard Matrix:")

print(np.array([[1,1],[1,-1]])/np.sqrt(2))

print()

print("Z Matrix:")

print(np.array([[1,0],[0,-1]]))

print()

print("X Matrix:")

print(np.array([[0,1],[1,0]]))

print()

H = np.array([[1,1],[1,-1]])/np.sqrt(2)

Z = np.array([[1,0],[0,-1]])

HZH = H @ Z @ H

print("HZH Matrix:")

print(HZH)

print()

print("Is HZH equal to X?")

print(np.allclose(HZH,np.array([[0,1],[1,0]])))
```
