```python
import numpy as np
import matplotlib.pyplot as plt

lengths = list(range(2, 11))

simon_queries = []
classical_queries = []
speedups = []

for n in lengths:
    simon = n
    classical = 2 ** (n - 1)

    simon_queries.append(simon)
    classical_queries.append(classical)

    speedups.append(classical / simon)

print("SIMON VS CLASSICAL BRUTE FORCE")
print("=" * 70)

print(
    f"{'n':<8}"
    f"{'Simon Queries':<20}"
    f"{'Classical Queries':<22}"
    f"{'Speedup':<15}"
)

print("-" * 70)

for n, s, c, speedup in zip(
    lengths,
    simon_queries,
    classical_queries,
    speedups
):
    print(
        f"{n:<8}"
        f"{s:<20}"
        f"{c:<22}"
        f"{speedup:.2f}x"
    )

plt.figure(figsize=(9, 5))

plt.plot(
    lengths,
    simon_queries,
    marker="o",
    label="Simon's Algorithm"
)

plt.plot(
    lengths,
    classical_queries,
    marker="s",
    label="Classical Brute Force"
)

plt.xlabel("Secret String Length (n)")
plt.ylabel("Oracle Queries")
plt.title("Simon Algorithm vs Classical Brute Force")

plt.yscale("log")

plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()

plt.figure(figsize=(9, 5))

plt.plot(
    lengths,
    speedups,
    marker="o"
)

plt.xlabel("Secret String Length (n)")
plt.ylabel("Classical / Simon Query Ratio")
plt.title("Speedup of Simon's Algorithm")

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
