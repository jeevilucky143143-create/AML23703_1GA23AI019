```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(2)
qc.h(0)
qc.h(1)

print("Quantum Circuit:\n")
print(qc.draw())

state = Statevector.from_instruction(qc)

print("\nStatevector:")
print(state)

print("\nProbability Amplitudes:")
labels = ["|00⟩","|01⟩","|10⟩","|11⟩"]

for label, amplitude in zip(labels, state.data):
    print(f"{label} : {amplitude}")

norm = sum(abs(amplitude)**2 for amplitude in state.data)

print(f"\nSum of Squared Amplitudes = {norm:.4f}")

if abs(norm-1.0) < 1e-10:
    print("Verification: PASS (Statevector is normalized)")
else:
    print("Verification: FAIL")
```
